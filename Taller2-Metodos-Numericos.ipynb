### Objetivo 1- Gráficar la funcion  
* Vector xs
* Definir funcion ()

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

xs = list(np.linspace(0, 5, 50))
```

```python
def f(x):
    _y = x**3 - 3 * x**2 + x - 1
    print(f"{x:2f}, {_y: .2f}")
    return _y
```

```python
ys = [f(x) for x in xs]
```

```python
import matplotlib.pyplot as plt

plt.plot(xs, ys, "o:")
plt.xlabel("xs")
plt.ylabel("ys")
plt.title("Gráfica de ys vs xs")
plt.grid(True)
plt.show()
```
